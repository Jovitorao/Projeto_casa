### 7 Indentificação de Outliers


```python
fig_hist = px.histogram(
    df_traduzido,
    x = 'preco',
    nbins= 100,
    histnorm= 'probability density',
    title= 'Distribuição dos Preços'
)
fig_hist.show()
```

```python
df_traduzido['log_preco'] = np.log(df_traduzido['preco'])

fig_hist_log = px.histogram(
    df_traduzido,
    x = 'log_preco',
    nbins=100,
    histnorm='probability density',
    title= 'Distribuição do log(preco)'
)
fig_hist_log.show()
```

```python
fig_box= px.box(df_traduzido,
                y='preco',
                title = 'Distribuição de Preços com outliers',
                points='outliers'
                )
fig_box.show()
```

```python
#Mais eficiente para outliers maiores do que a mediana

Q1 = df_traduzido['preco'].quantile(0.25)
Q3 = df_traduzido['preco'].quantile(0.75)

IQR = Q3-Q1

limite_inferior = Q1 - 1.5*IQR
limite_superior = Q3 +1.5*IQR


print(f'Q1: {Q1:,.0f}')
print(f'Q3: {Q3:,.0f}')
print(f'IQR: {IQR:,.0f}')
print(f'Limite superior: {limite_superior:,.0f}')
print(f'Limite inferior: {limite_inferior:,.0f}')

df_traduzido['classificacao']  = df_traduzido['preco'].apply(
    lambda x: 'Outlier' if x > limite_superior or x < limite_inferior else 'Normal'
)

total = len(df_traduzido)
n_outliers = (df_traduzido['classificacao'] == 'Outlier').sum()
print(f'\nOutliers encontrados: {n_outliers} ({n_outliers/total*100:.1f}%)')
```

```python
fig4 = px.scatter_mapbox(
    df_traduzido,
    lat='latitude',
    lon='longitude',
    color=df_traduzido['preco'].apply(
        lambda x: 'Outlier' if x > limite_superior else 'Normal'
    ),
    color_discrete_map={'Outlier': 'red', 'Normal': 'steelblue'},
    size='preco',
    zoom=9,
    mapbox_style='carto-positron',
    title='Localização Geográfica dos Outliers de Preço'
)
fig4.show()
```

```python
#Utilizando o z score com,og para identificar os outliers
media_log = df_traduzido['log_preco'].mean()
desvio_log= df_traduzido['log_preco'].std()

# (valor - média) / desvio padrãoárea
# np.abs() pega o valor absoluto 

df_traduzido['z_score'] = (df_traduzido['log_preco'] - media_log)/desvio_log
df_traduzido['z_score_abs'] = df_traduzido['z_score'].abs()

df_traduzido['classificacao_z'] = df_traduzido['z_score_abs'].apply(
    lambda z: 'Outlier' if z > 3 else 'Normal'
)
n_outliers_z = (df_traduzido['classificacao_z'] == 'Outlier').sum()
print(f'Outliers pelo Z-Score: {n_outliers_z} ({n_outliers_z/total*100:.1f}%)')
```

```python
outliers = df_traduzido[df_traduzido['classificacao_z'] == 'Outlier']
normais  = df_traduzido[df_traduzido['classificacao_z'] == 'Normal']

total = len(df_traduzido)
n_out = len(outliers)

print(f'Total de imóveis:  {total}')
print(f'Outliers:          {n_out} ({n_out/total*100:.1f}%)')
print(f'Normais:           {len(normais)}')
print(f'\nZ-score máximo:    {df_traduzido["z_score_abs"].max():.2f}')
print(f'Z-score mínimo:    {df_traduzido["z_score_abs"].min():.2f}')
print(f'\nPreço mais alto entre outliers:  {outliers["preco"].max():,.0f}')
print(f'Preço mais baixo entre outliers: {outliers["preco"].min():,.0f}')
```

```python
fig = px.box(
    df_traduzido,
    y='preco',
    color='classificacao_z',
    color_discrete_map={'Outlier': 'red', 'Normal': 'steelblue'},
    title='Boxplot de Preços — Outliers pelo Z-Score',
    labels={'preco': 'Preço', 'outlier_z': 'Classificação'},
    points='outliers'   # mostra só os pontos que são outliers, não todos
)

fig.show()
```
