## 🔹 FASE 2 — Consultas e Agregações

### 2. Análise Histórica de Preços

```python
df_traduzido['ANO'] = df_traduzido['data'].dt.year
df_traduzido['MES'] = df_traduzido['data'].dt.month
```

```python
df_traduzido.head()
```

```python
media_preco_ano = df_traduzido.groupby('ANO')['preco'].mean().round(2).reset_index().rename(columns={'preco' : 'Média de Preço'})
media_preco_ano
```

```python
media_mensal_geral = df_traduzido.groupby(['ANO','MES'])['preco'].mean().reset_index().round(2)
media_mensal_geral
```

```python
#Gráfico de linha no periodo do datset 
media_mensal_geral['data_referencia'] = pd.to_datetime(media_mensal_geral['ANO'].astype(str)+ '-' +media_mensal_geral['MES'].astype(str) + '-01')

fig = px.line(media_mensal_geral , x ='data_referencia', y = 'preco', color='ANO', color_discrete_map={2014:'blue' , 2015:'red'} 
              , title = 'Comportamento de Preços: 2014 (Azul) vs 2015 (Vermelho)', markers =True)

fig.update_xaxes(
    dtick = 'M1',
    tickformat='%b\n%Y',
    ticklabelmode="period"

)
fig.show()
```
